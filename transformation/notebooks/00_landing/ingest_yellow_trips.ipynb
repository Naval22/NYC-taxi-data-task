### Landing: download yellow taxi trips for March 2025 into `raw/`



```python
%run ../../config/config
```

```python
import os
import shutil
import urllib.request

try:
    if os.path.exists(TRIPS_FILE) and os.path.getsize(TRIPS_FILE) > 0:
        print(f"File already in raw/: {TRIPS_FILE}")
    else:
        os.makedirs(RAW_DIR, exist_ok=True)
        req = urllib.request.Request(TRIPS_URL, headers={"User-Agent": "Mozilla/5.0"})
        with urllib.request.urlopen(req) as response, open(TRIPS_FILE, "wb") as f:
            shutil.copyfileobj(response, f)
        print(f"File successfully downloaded: {TRIPS_FILE}")
    dbutils.jobs.taskValues.set(key="continue_downstream", value="yes")
except Exception as e:
    dbutils.jobs.taskValues.set(key="continue_downstream", value="no")
    raise Exception(f"Download failed: {e}")
```
