### Landing: download the taxi zone lookup CSV into `raw/`

```python
%run ../../config/config
```

```python
import os
import shutil
import urllib.request

try:
    if os.path.exists(ZONES_FILE) and os.path.getsize(ZONES_FILE) > 0:
        print(f"File already in raw/: {ZONES_FILE}")
    else:
        os.makedirs(RAW_DIR, exist_ok=True)
        req = urllib.request.Request(ZONES_URL, headers={"User-Agent": "Mozilla/5.0"})
        with urllib.request.urlopen(req) as response, open(ZONES_FILE, "wb") as f:
            shutil.copyfileobj(response, f)
        print(f"File successfully downloaded: {ZONES_FILE}")
    dbutils.jobs.taskValues.set(key="continue_downstream", value="yes")
except Exception as e:
    dbutils.jobs.taskValues.set(key="continue_downstream", value="no")
    raise Exception(f"Download failed: {e}")
```
