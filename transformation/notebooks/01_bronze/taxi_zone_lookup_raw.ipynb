### Bronze: `01_bronze.taxi_zone_lookup_raw` — lookup CSV as delivered (all columns as strings)

```python
from pyspark.sql.functions import current_timestamp

# Spark keeps the literal "N/A" as a string (only empty fields become NULL), which matters for zones 264/265.
df = spark.read.format("csv").option("header", True).load(ZONES_FILE) \
          .withColumn("processing_timestamp", current_timestamp())

```

```python
#overwrite data
df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(tbl("01_bronze", "taxi_zone_lookup_raw"))
```
